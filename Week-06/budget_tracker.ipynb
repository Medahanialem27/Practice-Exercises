# FinTrack - Personal Budget Tracker
**Course:** Data Analytics | **Date:** November 16 2026
This notebook collects monthly income and expenses, then creates a formatted budget summary.

```python
analyst = input("Analyst name:")
month = input("Budget month:")
```

## Section 1: Analyst information
Netsanet Aweke, November 2026
## Section 2: Monthly income and expenses


```python
income = float(input("Monthly income $: "))
rent = float(input("Rent / housing $: "))
food = float(input("Food / groceries $: "))
transport = float(input("Transport $: "))
utilities = float(input("utilities $: "))          
```

## Section 3: Budget calculations
This section calculates total expenses, the remaining budget, and the saving rate

```python
total_expenses = rent + food + transport + utilities
remaining = income - total_expenses
saving_rate = (remaining / income) * 100
```

```python
print(f"Total expenses: ${total_expenses:,.2f}")
print(f"Remaining: ${remaining:,.2f}")
print(f"Saving rate: {saving_rate:.1f}%")
```

## Section 4: Budget summary report
The report below shows the analyst, month, income, expenses, and savings rate.

```python
print(f"""
====== FinTrack Budget Report ======
Analyst : {analyst}
Month : {month}
------------------------------------
Income : ${income:,.2f}
Rent/Housing : ${rent:,.2f}
Food/Groceries : ${food:,.2f}
Transport : ${transport:,.2f}
Utilities : ${utilities:,.2f}
------------------------------------
Total expenses : ${total_expenses:,.2f}
Remaining : ${remaining:,.2f}
Savings rate: {saving_rate:.1f}%
====================================
""")
```
